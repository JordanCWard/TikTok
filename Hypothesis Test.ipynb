# **Hypothesis Testing**

### Comparing Video Views: Verified vs. Unverified TikTok Accounts

This analysis investigates whether there is a significant difference in video view counts between verified and unverified TikTok accounts. Two research questions are addressed:

1. Do videos from verified and unverified accounts have different average view counts?
2. Is there a relationship between account verification status and the associated videos' view counts?

A two-sample t-test is used to evaluate these questions at a significance level of 5% (α = 0.05).

## Import Libraries

```python
# Import necessary libraries
import pandas as pd  # For data manipulation and analysis
from scipy import stats  # For statistical tests
```

## Load Data

```python
# Load dataset into a pandas DataFrame
data = pd.read_csv("tiktok_dataset.csv")
```

## **Data exploration**

```python
# Preview the first few rows of the dataset
data.head()
```

```python
# Check for missing values
data.isnull().sum()
```

```python
# Drop rows with missing values
data = data.dropna()
```

## Data Preparation
Filter the dataset based on account verification status.

```python
# Filter video view counts for unverified accounts
not_verified = data[data["verified_status"] == "not verified"]["video_view_count"]

# Filter video view counts for verified accounts
verified = data[data["verified_status"] == "verified"]["video_view_count"]
```

## **Hypothesis testing**

*   $H_0$: Verified and unverified TikTok accounts have no difference in video views; any observed difference is due to chance.
*   $H_A$: Verified and unverified TikTok accounts differ in video views; the observed difference reflects a true population difference.

Statistical significance was assessed at the 5% level (α = 0.05) using a two-sample t-test.

```python
# Two-sample t-test to compare means
stats.ttest_ind(a=not_verified, b=verified, equal_var=False)
```

## Results and Interpretation

Since the p-value is extremely small (much smaller than the significance level of 5%), reject the null hypothesis. Conclude that there is a statistically significant difference in the mean video view count between verified and unverified accounts on TikTok.

Since the p-value is extremely small (p < 0.05), we reject the null hypothesis. This indicates a statistically significant difference in the mean video view counts between verified and unverified TikTok accounts.

The analysis suggests that verified and unverified accounts differ meaningfully in their average view counts, which may reflect underlying behavioral or structural differences between the two groups. Further investigation could explore potential drivers of this disparity. For instance, do unverified accounts more frequently employ clickbait strategies, or are they more likely to be associated with spam activity that artificially inflates view counts?
